## 1. PEAS Model

**PEAS** = Performance measure, Environment, Actuators, Sensors

### Problem 1: Food-Waste Prediction
| Component | Description |
|-----------|-------------|
| **Performance** | Accurate kg prediction, low MAE, reduced waste |
| **Environment** | Kitchen/cafeteria context: people count, meals, day type, history |
| **Actuators** | Display prediction, alert staff, adjust procurement |
| **Sensors** | Input form fields, historical waste records |

### Problem 2: Waste-Segregation Decision Support
| Component | Description |
|-----------|-------------|
| **Performance** | Correct bin recommendation, contamination reduction |
| **Environment** | Waste item description and condition (e.g., soiled or clean) |
| **Actuators** | Bin recommendation display |
| **Sensors** | User-entered item name and condition flags |

## 2. Problem Formulation

### Food-Waste Prediction
- **Initial state:** No prediction yet; inputs are blank.
- **States:** Combinations of (people, meals, day_type, hist_waste).
- **Actions:** Train model, make prediction, evaluate.
- **Transition model:** Feeding inputs through the trained model produces a prediction.
- **Goal test:** Prediction error (MAE) below an acceptable threshold.
- **Cost:** Computational cost of training; cost of over/under-estimating waste.

### Waste-Segregation Decision Support
- **Initial state:** Unknown item category.
- **States:** Possible categories {wet, dry, residual}.
- **Actions:** Apply rules based on item and condition.
- **Transition model:** Rules map (item, condition) -> category.
- **Goal test:** Correct category assigned.
- **Cost:** Misclassification leads to contamination or wasted recyclables.
